```python
# Check whether duplicate feature rows have conflicting labels

df = X.copy()
df["target"] = y.iloc[:, 0].values

duplicate_groups = (
    df.groupby(list(X.columns))["target"]
    .nunique()
)

conflicting_duplicates = (duplicate_groups > 1).sum()

print("Duplicate feature groups:", X.duplicated().sum())
print("Duplicate groups with conflicting labels:", conflicting_duplicates)
```

```python
print("Original samples:", len(X))
print("Unique samples:", len(X.drop_duplicates()))
```

```python
# Combine features and target into one DataFrame

df = X.copy()
df["target"] = y.iloc[:, 0].values

print("Original dataset shape:", df.shape)
```

```python
# Count how many different labels exist for each feature combination

label_counts = (
    df.groupby(list(X.columns))["target"]
    .value_counts()
    .unstack(fill_value=0)
)

# Find feature combinations that have BOTH labels
conflicting_groups = label_counts[
    (label_counts > 0).sum(axis=1) > 1
]

print("Conflicting feature groups:", len(conflicting_groups))
```

```python
display(conflicting_groups.head(10))
```

```python
# Count how many original rows belong to conflicting groups

conflicting_index = conflicting_groups.index

conflicting_mask = df.set_index(list(X.columns)).index.isin(conflicting_index)

print("Rows belonging to conflicting groups:",
      conflicting_mask.sum())
```

```python
# Remove rows belonging to conflicting feature groups

clean_df = df.loc[~conflicting_mask].copy()

print("Original rows:", len(df))
print("Rows removed:", len(df) - len(clean_df))
print("Rows remaining:", len(clean_df))
```

```python
# Check remaining duplicates

print("Duplicates remaining:",
      clean_df.drop(columns="target").duplicated().sum())
```

```python
# Remove exact duplicate feature rows

clean_df = clean_df.drop_duplicates(
    subset=list(X.columns),
    keep="first"
).copy()

print("Rows after deduplication:", len(clean_df))
```

```python
print("Final dataset shape:", clean_df.shape)

print("\nRemaining duplicates:",
      clean_df.drop(columns="target").duplicated().sum())

print("\nMissing values:",
      clean_df.isnull().sum().sum())
```

```python
from sklearn.linear_model import LogisticRegression

print("Logistic Regression imported successfully!")
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    pos_label=-1
)

recall = recall_score(
    y_test,
    y_pred,
    pos_label=-1
)

f1 = f1_score(
    y_test,
    y_pred,
    pos_label=-1
)

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
```
